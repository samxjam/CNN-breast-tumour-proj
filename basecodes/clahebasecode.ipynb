```python
import os
import cv2
import shutil

input_dataset = "D:\Documents\DATA3888_Assignment\Cells\OversamplingData"  # original dataset folder
output_dataset = "D:\Documents\DATA3888_Assignment\Cells\clahe_dataset"     # new dataset folder for CLAHE images

clip_limit = 3.0
tile_grid_size = (8, 8)

def apply_clahe(img):

    lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
    l, a, b = cv2.split(lab)

    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
    cl = clahe.apply(l)

    merged = cv2.merge((cl, a, b))
    final_img = cv2.cvtColor(merged, cv2.COLOR_LAB2RGB)
    
    return final_img

def process_dataset(input_dir, output_dir):
    if os.path.exists(output_dir):
        shutil.rmtree(output_dir)
    os.makedirs(output_dir)

    for root, dirs, files in os.walk(input_dir):
        for file in files:
            if file.lower().endswith((".png")):
                input_path = os.path.join(root, file)

                relative_path = os.path.relpath(root, input_dir)
                output_folder = os.path.join(output_dir, relative_path)
                os.makedirs(output_folder, exist_ok=True)
                output_path = os.path.join(output_folder, file)

                img = cv2.imread(input_path)
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 
                processed_img = apply_clahe(img)

                processed_img_bgr = cv2.cvtColor(processed_img, cv2.COLOR_RGB2BGR)
                cv2.imwrite(output_path, processed_img_bgr)

    print(f"CLAHE dataset created at {output_dir}")

process_dataset(input_dataset, output_dataset)
```
