```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dlclahe = pd.read_csv("dlcnn_MS_CLAHE_history.csv")
dlnclahe = pd.read_csv("dlcnn_MS_NONCLAHE_history.csv")
mcclahe = pd.read_csv("multiclass_CLAHE_history.csv")
mcnclahe = pd.read_csv("multiclass_NONCLAHE_history.csv")
```

```python
def history_plot(history_df, metric="loss"):
    fig, ax = plt.subplots()
    ax.plot(history_df["binary_output_" + metric], 
            label=f"Train Binary {metric.capitalize()}", marker='o', lw=1.5, ms=4)
    ax.plot(history_df["val_binary_output_" + metric], 
            label=f"Val Binary {metric.capitalize()}", marker='o', lw=1.5, ms=4)
    ax.plot(history_df["multiclass_output_" + metric], 
            label=f"Train Multiclass {metric.capitalize()}", marker='o', lw=1.5, ms=4)
    ax.plot(history_df["val_multiclass_output_" + metric], 
            label=f"Val Multiclass {metric.capitalize()}", marker='o', lw=1.5, ms=4)
    ax.set_title(f"{metric.capitalize()} over Epochs")
    ax.set_xlabel("Epoch")
    ax.set_ylabel(metric.capitalize())
    ax.legend()
    plt.tight_layout()
    plt.grid(alpha=0.5)
```

```python
fig, ax = plt.subplots()

ax.plot(dlclahe["binary_output_loss"], 
        label="DL Train Binary Loss", marker='o', lw=1.5, ms=4)

ax.plot(dlclahe["val_binary_output_loss"], 
        label="DL Val Binary Loss", marker='o', lw=1.5, ms=4)

ax.plot(dlclahe["multiclass_output_loss"], 
        label="DL Train Multiclass Loss", marker='o', lw=1.5, ms=4)

ax.plot(dlclahe["val_multiclass_output_loss"], 
        label="DL Val Multiclass Loss", marker='o', lw=1.5, ms=4)

ax.plot(mcclahe["multiclass_output_loss"], 
        label="MC Train Loss", marker='o', lw=1.5, ms=4)

ax.plot(mcclahe["val_multiclass_output_loss"], 
        label="MC Val Loss", marker='o', lw=1.5, ms=4)

ax.set_title("Loss over Epochs")
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss")
ax.legend()
plt.tight_layout()
plt.grid(alpha=0.5)
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(dlclahe["binary_output_accuracy"], 
        label="DL Train Binary Accuracy", marker='o', lw=1.5, ms=4)

ax.plot(dlclahe["val_binary_output_accuracy"], 
        label="DL Val Binary Accuracy", marker='o', lw=1.5, ms=4)

ax.plot(dlclahe["multiclass_output_accuracy"], 
        label="DL Train Multiclass Accuracy", marker='o', lw=1.5, ms=4)

ax.plot(dlclahe["val_multiclass_output_accuracy"], 
        label="DL Val Multiclass Accuracy", marker='o', lw=1.5, ms=4)

ax.plot(mcclahe["multiclass_output_accuracy"], 
        label="MC Train Accuracy", marker='o', lw=1.5, ms=4)

ax.plot(mcclahe["val_multiclass_output_accuracy"], 
        label="MC Val Accuracy", marker='o', lw=1.5, ms=4)

ax.set_title("Accuracy over Epochs")
ax.set_xlabel("Epoch")
ax.set_ylabel("Accuracy")
ax.legend()
plt.tight_layout()
plt.grid(alpha=0.5)
plt.show()
```
